## Are the data suitable for answering the question?

The four flags need three measurements: precipitation (PRCP), snowfall (SNOW), and the daily low temperature (TMIN). First, how many non-null values does each column have?

```python
df_chicago.notna().sum()
```

The columns fall into three groups:

| Columns | What they are | Decision |
|---|---|---|
| PRCP, SNOW, TMIN | Daily precipitation, snowfall, and low temperature | Keep, these drive the flags |
| DAPR, MDPR | Multi-day precipitation totals | Drop, these would double count the daily PRCP values |
| SNWD | Snow depth on the ground | Drop, difficult to judge the impact - could be snow left over from and earlier day, it is unclear if the roads are plowed, etc |
| TAVG | Average temperature | Drop, it has no values at all |
| TMAX, TOBS | High temperature and temperature at observation time | Drop, TMIN already answers the freezing question and TOBS is only reported in very few rows |
| WT01 to WT11 | Weather type indicators like fog and thunder | Drop, they are reported in very few rows |
| NAME | Station name | Drop, STATION already identifies each station and neither is needed after the stations are combined |

##### How many stations report each of the three measurements on a given day?

The flags are daily, so every day needs at least one reading of each measurement. Temperature is only reported by a small share of rows, so this is the one to watch.

```python
per_day = df_chicago.groupby('DATE').agg(
    prcp_stations = ('PRCP', 'count'),
    snow_stations = ('SNOW', 'count'),
    tmin_stations = ('TMIN', 'count')
)

per_day.describe().loc[['min', 'mean', 'max']].round(1)
```

Every day has at least 97 stations reporting precipitation, 25 reporting snowfall, and 7 reporting a low temperature. No day is missing a measurement, so the data are suitable and there is nothing to impute.

## Consolidate the stations into one row per day

The weather is spread across many stations, but the rideshare data only needs one weather value per day for the whole city. I took the mean across stations for each measurement.

- mean() skips missing values, so a station only counts on the days it actually reported.
- The mean describes the city as a whole. A thunderstorm over one suburb will not make the entire city a heavy rain day.
- It also dampens a single station with a bad reading.

```python
daily = (
    df_weather
    .groupby('DATE')
    .agg(
        prcp = ('PRCP', 'mean'),
        snow = ('SNOW', 'mean'),
        tmin = ('TMIN', 'mean')
    )
    .reset_index()
)
```

### Rename columns to follow best practices

Rename `DATE` to be lowercase so it matches the other column names.

```python
daily = daily.rename(columns = {'DATE': 'date'})
```

## Choose thresholds

The flags need cutoffs, and these are judgment calls. I put them in one place so they can be changed and the notebook rerun.

| Flag | Rule | Reasoning |
|---|---|---|
| light_rain | average precipitation from 0.01 in up to 0.50 in | 0.01 in is the smallest amount the gauges measure, so it separates a real rain day from a dry one |
| heavy_rain | average precipitation of 0.50 in or more | Half an inch in a day is a common cutoff for heavy rain. Because this is an average over many stations, it is a fairly strict bar |
| snow | average snowfall of 0.10 in or more | Requiring a small average keeps one station's trace report from labeling the whole city as a snow day |
| below_freezing | average daily low under 32 °F | The low is the coldest the day got. Using the high would only flag days that never got above freezing |

```python
LIGHT_RAIN_MIN = 0.01   # inches per day
HEAVY_RAIN_MIN = 0.50   # inches per day
SNOW_MIN = 0.10         # inches per day
FREEZING_F = 32         # degrees Fahrenheit
```

##### How sensitive is the result to the heavy rain cutoff?

The heavy rain cutoff changes the result the most, so here is how many days each cutoff would reach (before removing snow days).

```python
cutoffs = [0.25, 0.50, 0.75, 1.00]

pd.Series(
    [int((daily['prcp'] >= c).sum()) for c in cutoffs],
    index = [f'{c:.2f} in' for c in cutoffs],
    name = 'days_at_or_above'
).to_frame()
```

## Create the true/false flags

PRCP includes snow melted into water. On a snow day the precipitation would be counted as rain, so I only allow the rain flags to be True when it was not a snow day.

```python
# a snow day is any day with enough average snowfall
snow_day = daily['snow'] >= SNOW_MIN
```

```python
weather = pd.DataFrame({
    'date': daily['date'],

    # light rain: measurable but under the heavy cutoff, and not a snow day
    'light_rain': (
        (daily['prcp'] >= LIGHT_RAIN_MIN)
        & (daily['prcp'] < HEAVY_RAIN_MIN)
        & ~snow_day
    ),

    # heavy rain: at or above the heavy cutoff, and not a snow day
    'heavy_rain': (
        (daily['prcp'] >= HEAVY_RAIN_MIN)
        & ~snow_day
    ),

    'snow': snow_day,

    # the average daily low was under freezing
    'below_freezing': daily['tmin'] < FREEZING_F
})
```

## Check the result

##### Check the data types and number of missing values

```python
weather.info()
```

##### How many days are flagged for each condition?

```python
weather.drop(columns = 'date').sum()
```

The flags should make sense together. Light rain and heavy rain should never both be True, and rain should never be flagged on a snow day.

```python
assert not (weather['light_rain'] & weather['heavy_rain']).any()
assert not ((weather['light_rain'] | weather['heavy_rain']) & weather['snow']).any()

print('Checks passed')
```

##### Does snow fall on days below freezing?

Snow should mostly fall on days with a low under freezing.

```python
weather.loc[weather['snow'], 'below_freezing'].mean()
```
